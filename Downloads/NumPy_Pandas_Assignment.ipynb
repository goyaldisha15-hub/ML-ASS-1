```python
#np.array()

#Description: creates a n-dimensional array from python sequences

#1 1-D array from a list
arr1 = np.array([1,2,3,4,5])
print(arr1)

#2 2-D array from a list 
arr2 = np.array([[1,2,3],[4,5,6]])
print(arr2)

#3 1-D array from a Tuple
arr3 = np.array((5, 10, 15, 20))
print(arr3)
```

```python
#np.zeros()

#1 creating a 1D array of 5 zeroes
arr1 = np.zeros(5)
print(arr1)

#2  creating a 2D array of 2 rows 3 columns of zeroes
arr2 = np.zeros((2, 3))
print(arr2)

#3  creating a 3D array of 2 matrices,2rows,2columns of zeroes 
arr3 = np.zeros((2, 2, 3))
print(arr3)
```

```python
#np.ones()
# 1. Creating a 1D array of 5 ones (float by default)
arr1 = np.ones(5)

# 2. Creating a 2D array (2 rows, 3 columns) of ones
arr2 = np.ones((2, 3))

# 3. Creating a 2D integer array (3 rows, 2 columns) of ones
arr3 = np.ones((3, 2), dtype=int)
```

```python
arr=np.array([[12,3,6],[4,335,64]])
print(arr.size) #total size
print(arr.ndim)#check dimension
```

```python
#sum operations
arr=np.array([[12,3,6],[4,335,64]])
#1 give sum of whole array
print(arr.sum())
#2 add vertically (column wise)
print(arr.sum(axis=0))
#3 add horizontally (row wise)
print(arr.sum(axis=1)) 
```

```python
#reshape operations
arr = np.array([[12, 3, 6], [4, 335, 64]])

# 1. Reshape automatically specifying rows using -1
reshape1 = arr.reshape(-1, 3)

# 2. Reshape into a 3D array (depth=1, rows=2, cols=3)
reshape2 = arr.reshape(1, 2, 3)

# 3. Reshape into a 1D vector (6 elements)
reshape3 = arr.reshape(6)
```

```python
#ravel operation
# 1. converting 2D array into 1D array
arr1 = np.array([[1, 2, 3], [4, 5, 6]])
print(arr1.ravel())


# 2.chnaging value(original array change ho jata hai)
arr2 = np.array([[12, 3, 6], [4, 335, 64]])
demo = arr2.ravel()
demo[0] = 100  # changes first element
print(arr2)

# 3. converting 3D into 1D
arr3 = np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])
print(arr3.ravel())
```

```python
arr=np.array([12,3,6,3,56,78,4,335,64,65,78,97])
#eye= gives identity matrix
k=arr.reshape(3,4)
arr[0]=100
k
#change in original array
```

```python
#copy operation
arr = np.array([12, 3, 6, 3, 56, 78, 4, 335, 64, 65, 78, 97])
k = arr.reshape(3, 4).copy()
arr[0] = 100  # will not change copy array
print(k)
```

```python
#broadcasting
d = np.array([100, 56, 87, 56, 65, 78, 89, 76, 98, 45, 57, 87])
print(d % 10)  # numpy array modulo operation
print(d * 2)  # multiply each element by 2
print(d + 10)  # add 10 to each element
```

```python
#masking
d=np.array([100,23,45,67,89,98,87,65,43,32,56,78])
mask = (d>80)
print(mask)
print(d[mask]) #will print true values
print(d[~mask]) #approximation symbol prints false values
```

```python
d=np.array([100,23,45,67,89,98,87,65,43,32,56,78])
m=(d>70)&(d<90)
print(d[m])
```

```python
# conditional selection (np.where)
d = np.array([100, 23, 45, 67, 89, 98, 87, 65, 43, 32, 56, 78])

# 1. Return index of the element which satisfies condition (greater than 70)
print(np.where(d > 70))  # return index of elements > 70

# 2. Return index of the element which is even number
print(np.where(d % 2 == 0))  # return index of even elements

# 3. Return index of the element which is less than or equal to 50
print(np.where(d <= 50))  # return index of elements <= 50
```

```python
#np.arange()
print(np.arange(10))  # 0 to 9 numbers
print(np.arange(5, 20, 2))  # start, stop, step
print(np.arange(10, 0, -1))  # countdown sequence
```

```python
#np.linspace()
print(np.linspace(0, 1, 5))  # 5 linearly spaced elements
print(np.linspace(10, 50, 5))  # 10 to 50 split into 5 parts
print(np.linspace(1, 10, 3))  # 3 elements between 1 and 10
```

```python
#np.eye() used for identity matrix
print(np.eye(3))  # 3x3 identity matrix
print(np.eye(2, 4))  # 2x4 rectangular identity matrix
print(np.eye(3, dtype=int))  # integer identity matrix
```
